#What is the dictionaries?

A dictionary is a collection of key-value pairs. It's used to store data where each item has a unique identifier (called the key) and an associated value.

- Think of it like a real dictionary:

   1. Word = key

   2. Definition = value

Dictionaries: Key-value pairs, unordered, mutable, keys must be unique and immutable

```python
myDict = {
    "name": "Shamil",
    "username": "Shamil2007",
    "email": "sherifzade09@gmail.com"
}
myDict
```

```python
myDict = dict(name="Shamil", username="Shamil2007", email="sherifzade09@gmail.com")
myDict
```

#Access a value

#Get all keys/values/items

```python
myDict.keys()
```

```python
myDict.values()
```

```python
myDict.items()
```

```python
for key in myDict: # equivalent to myDict.keys()
    print(key)
```

```python
for value in myDict.values():
    print(value)
```

```python
counter = 1
for key, value in myDict.items():
    print(f"{counter}. Key: {key}, value: {value}")
    counter += 1
```

#Counting Frequencies in an Array Using a Dictionary

```python
fruits = ["apple", "banana", "apple", "orange", "banana", "apple", "pear", "orange", "banana", "apple"]
```

```python
count = {}
for fruit in fruits:
    if fruit in count:
        count[fruit] += 1
    else:
        count[fruit] = 1

count
```

```python
from collections import Counter

count = Counter(fruits)
count
```
