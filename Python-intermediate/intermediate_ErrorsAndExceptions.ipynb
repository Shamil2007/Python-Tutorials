# What Are Errors and Exceptions?


In programming, errors are problems that stop the normal flow of a program. Python uses a system called exceptions to handle these problems gracefully.

There are two main types:
1. Syntax Errors: Mistakes in code structure, caught before running
2. Problems that occur while running the program

# Syntax Erorrs (Compile Time)

These happen when your code is written incorrectly and won’t even run.

```python
print("Hello World"  # ❌ Missing closing parenthesis
```

```python
x = 5
if 5 > 0:
    print(f"{x} is positive")
else
    print(f"{x} is negative")
```

#2. Exceptions (Run-Time Errors)


Even if your syntax is correct, things can go wrong while the program is running.

```python
a = 5 / 0  # ❌ ZeroDivisionError
```

# Common Type of Exceptions

| Exception           | When It Happens                           |
| ------------------- | ----------------------------------------- |
| `ZeroDivisionError` | Dividing by 0                             |
| `TypeError`         | Wrong data type (e.g., `1 + "a"`)         |
| `ValueError`        | Invalid value (e.g., `int("abc")`)        |
| `IndexError`        | Invalid list index                        |
| `KeyError`          | Missing dictionary key                    |
| `FileNotFoundError` | File doesn’t exist                        |
| `ImportError`       | Can't find a module                       |
| `AttributeError`    | Accessing a non-existing method/attribute |


Example of Error Flow

```python
print("Starting...")
x = int("hello")  # This will raise ValueError
print("Done")
```

The program crashes and never reaches print("Done").

# Try-Except: Handling Exceptions


Use try-except to prevent the program from crashing:

```python
try:
    x = int("hello")
except:
    print("Oops! That was not a number.")
```

However if you want to be more specific

```python
try:
    x = int("hello")
except ValueError as VE:
    print(f"Oops! That was not a number. Error is {VE}")
```

```python
num1 = 10
num2 = 0
try:
    print(num1 / num2)
except ZeroDivisionError as Z:
    print(f"Error is {Z}")
```

```python
try:
    # code that might raise an exception
except SomeError:
    # code to run if there's an error
except SomeError2:
    # code to run if there's a another error
else:
    # code to run if no error happened
finally:
    # always runs (clean-up, closing files, etc.)
```

# Example with All:

```python
try:
    number = int(input("Enter a number: "))
except SyntaxError:
    print("There's Syntax error!")
except ValueError:
    print("Invalid number!")
else:
    print(f"You entered {number}")
finally:
    print("Program finished.")
```

#Raising Your Own Exceptions

You can use raise to throw errors on purpose:

```python
def set_age(age):
    if age < 0:
        raise ValueError("Age cannot be negative!")

set_age(-5)
```

#Custom Exceptions

```python
class MyCustomError(Exception):
    pass

raise MyCustomError("Something went wrong!")
```

#What is assert?

assert is a debugging tool that helps you test assumptions in your code. It checks a condition, and if that condition is False, it raises an AssertionError and stops the program.

Syntax:
        
    assert condition, "Error message if condition is False"


- condition: any expression that should be True.

- "Error message" (optional): shown when assertion fails.

```python
x = 10
assert x > 0, "x must be positive"
print("x is positive")
```

```python
x = -5
assert x > 0, "x must be positive"
print("x is positive")
```
