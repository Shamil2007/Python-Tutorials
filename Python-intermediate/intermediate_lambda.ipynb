#What is a lambda function?

- A lambda function is a small, anonymous function defined without a name.
- It's used for short, simple functions that are created where they are needed — usually passed as arguments.

Syntax:

    lambda arguments: expression

You can think of it like a one-line version of a def function.

```python
def square(x):
    return x * x
```

With lambda:

```python
square = lambda x: x * x
square(5)
```

```python
add10 = lambda num: num + 10
add10(12)
```

#Multiple Arguments

```python
add = lambda a, b: a + b
add(3, 4)
```

```python
fullName = lambda name, surname: name + " " + surname
fullName("Shamil", "Sharifzade")
```

#No arguments

```python
greet = lambda: "Hello!"
print(greet())
```

#Map function

Apply a function to every item in a list.
        
        map(func, seq)

```python
numbers = [2, 42, 12, 8, 53, 24]
squares = map(lambda num: num * num, numbers)
squares
```

```python
list(squares)
```

Instead of using map function, you can use comprehension syntax

```python
square = [num * num for num in numbers]
square
```

#Filter function

Filter out values from a list

        filter(func, seq)

```python
nums = [1, 2, 3, 4, 5, 7, 8, 10]
evens = filter(lambda x: x % 2 == 0, nums)
list(evens)
```

#Using lambda with sorted

Sort a list by a custom key

```python
words = ["apple", "kiwi", "banana"]
sorted(words)
```

```python
words = ["apple", "banana", "kiwi"]
sorted(words, key=lambda word: len(word))
```

This line sorts a list of words by the length of each word — from shortest to longest.

```python
sorted(words, key=lambda word: [-len(word)])
```

From longest to shortest
