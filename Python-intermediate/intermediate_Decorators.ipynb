#What is a Decorator?

A decorator is a function that modifies the behavior of another function — without changing its actual code.

You can think of it like:

- "Wrapping a function to extend its behavior."

Why Use Decorators?
1. Code reuse

2. Separation of concerns (e.g., logging, timing, authorization)

3. Clean and elegant syntax

Basic function:

```python
def greet():
    return "Hello"
```

Now, let’s say you want to add emojis before and after the greeting. You could edit the function — but decorators allow you to wrap it instead.

#Step 1: A Simple Decorator

```python
def decorator_function(original_function):
    def wrapper():
        print("Before the function runs")
        result = original_function()
        print("After the function runs")
        return result
    return wrapper
```

```python
@decorator_function
def say_hello():
    print("Hello!")

say_hello()
```

 The @decorator_function line is equivalent to:

```python
say_hello = decorator_function(say_hello)
say_hello()
```

#Decorators with Arguments

```python
def decorator_function(func):
    def wrapper(*args, **kwargs):
        print("Running with arguments")
        return func(*args, **kwargs)
    return wrapper

@decorator_function
def greet(name):
    print(f"Hello, {name}!")

greet("Shamil")
```

#Returning Values

```python
def double_result(func):
    def wrapper(*args, **kwargs):
        result = func(*args, **kwargs)
        return result * 2
    return wrapper

@double_result
def get_number():
    return 10

print(get_number())
```

#Built - in decorators

#1. @staticmethod

✅ Purpose:
- Defines a method that does not depend on instance or class.

- No self or cls.

📦 Use when:
- The function relates to the class logically but doesn’t need instance/class data.

```python
class Math:
    @staticmethod
    def add(a, b):
        return a + b

print(Math.add(3, 4))
```

#2. @classmethod

✅ Purpose:
- Passes the class itself (cls) as the first argument.

- Used to create factory methods or access/modify class variables.

📦 Use when:
- You need to access or modify class state, not just instance.

This example took from geeksforgeeks.org website

```python
class Geeks:
    course = 'DSA'
    list_of_instances = []

    def __init__(self, name):
        self.name = name
        Geeks.list_of_instances.append(self)

    @classmethod
    def get_course(cls):
        return f"Course: {cls.course}"

    @classmethod
    def get_instance_count(cls):
        return f"Number of instances: {len(cls.list_of_instances)}"

    @staticmethod
    def welcome_message():
        return "Welcome to Geeks for Geeks!"

# Creating instances
g1 = Geeks('Alice')
g2 = Geeks('Bob')

# Calling class methods
print(Geeks.get_course())
print(Geeks.get_instance_count())

# Calling static method
print(Geeks.welcome_message())
```

#3. @property

✅ Purpose:
- Makes a method accessible like an attribute.

- Helps to implement getter without ().

📦 Use when:
- You want to control access to an attribute (read-only, computed, etc.).

```python
class Circle:
    def __init__(self, radius):
        self._radius = radius

    @property
    def area(self):
        return 3.14 * self._radius ** 2

c = Circle(3)
print(c.area)
```

#4. @functools.lru_cache

✅ Purpose:
- Caches the result of expensive or repetitive function calls.

📦 Use when:
- You're calling a function multiple times with the same inputs (e.g. recursion, API calls).

```python
from functools import lru_cache

@lru_cache(maxsize=100)
def fib(n):
    if n < 2:
        return n
    return fib(n-1) + fib(n-2)

print(fib(100))
```

```python
print(fib(1000))
```

#5. @functools.wraps

✅ Purpose:
- Preserves the original function’s metadata (like name, docstring) when writing decorators.

📦 Use when:
- You're writing your own decorators.

```python
from functools import wraps

def logger(func):
    @wraps(func)
    def wrapper(*args, **kwargs):
        print(f"Running {func.__name__}")
        return func(*args, **kwargs)
    return wrapper

@logger
def hello():
    print("Hello!")

hello()
print(hello.__name__)  # 'hello', not 'wrapper'
```

#6. @abstractmethod

✅ Purpose:
- Used in abstract base classes.

- Forces subclasses to implement the method.

📦 Use when:
- You want to enforce a blueprint for subclasses.

```python
from abc import ABC, abstractmethod

class Animal(ABC):
    @abstractmethod
    def speak(self):
        pass

class Dog(Animal):
    def speak(self):
        return "Woof"
```

#7. @typing.final

✅ Purpose:
- Prevents method overriding in subclasses.

📦 Use when:
- You want to lock down a method or class from being modified.

```python
from typing import final

class Base:
    @final
    def important_method(self):
        print("Don't override me")

class Store(Base):
    def important_method(self): # Method "important_method" cannot override final method defined in class "Base"
        print("Store method")

store = Store()
store.important_method()
```

#8. @cached_property

✅ Purpose:
- Like @property, but caches the result — only computed once.

📦 Use when:
- You have an expensive computation you want to run only once per instance.

```python
from functools import cached_property

class Data:
    @cached_property
    def expensive_calculation(self):
        print("Calculating...")
        return sum(range(1000001))

d = Data()
print(d.expensive_calculation)  # Calculated
print(d.expensive_calculation)  # Cached
```
