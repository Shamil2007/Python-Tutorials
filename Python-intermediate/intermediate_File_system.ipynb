#What is a File Object?

A file object is Python's representation of an open file. It's an interface that allows you to read from or write to the underlying file. When you open a file, Python creates this object to manage all interactions with the file.

The core of Python's file I/O is the open() function, which returns a file object:

```python
file = open('filename.txt', 'r')  # Opens file for reading
```

The open() function takes several parameters:

Essentials:

1. filename: Path to the file
2. mode: How the file should be opened (read, write, append, etc.)
3. encoding: Text encoding

## 📂 Python File Modes

Python supports various file modes for handling files in **text** and **binary** formats.

---

### 📝 Text Modes:

| Mode  | Description                                 |
|-------|---------------------------------------------|
| `'r'` | Read (default). Fails if the file doesn't exist. |
| `'w'` | Write. Overwrites the file if it exists.     |
| `'a'` | Append. Writes data at the end of the file.  |
| `'x'` | Exclusive creation. Fails if file exists.     |
| `'r+'`| Read and write. File must exist.             |
| `'w+'`| Write and read. Truncates the file first.    |
| `'a+'`| Append and read. Creates file if not exists. |

---

### 🧱 Binary Modes:

Add `'b'` to any text mode to work with binary files:

| Mode   | Description                     |
|--------|---------------------------------|
| `'rb'` | Read binary                     |
| `'wb'` | Write binary                    |
| `'ab'` | Append binary                   |
| `'xb'` | Exclusive binary creation       |
| `'r+b'`| Read and write binary           |
| `'w+b'`| Write and read binary           |
| `'a+b'`| Append and read binary          |

---

✅ Use binary modes when dealing with non-text files such as images, audio files, etc.

#Context Managers (Recommended Approach)

The with statement ensures files are properly closed:

```python
with open('filename.txt', 'r') as file:
    content = file.read()
# File automatically closed here
```

#Reading Files

Reading entire file:


```python
with open('filename.txt', 'r') as f:
    content = f.read()  # Returns entire file as string

print(content)
```

Reading line by line:

```python
with open('filename.txt', 'r') as f:
    i = 1
    for line in f:  # Memory efficient
        print(f"Line {i}: {line.strip()}")
        i += 1
```

Reading all lines into a list:

```python
with open('filename.txt', 'r') as f:
    lines = f.readlines()  # Returns list of lines

print(lines)
```

Reading one line:

```python
with open('filename.txt', 'r') as f:
    first_line = f.readline()
    second_line = f.readline()

print(first_line)
print(second_line)
```

#Writing Files

Writing strings:

If the file doesn't exist, it will be created and texts will be written into it.

```python
with open('file.txt', 'w') as f:
    f.write('Hello, World!')
    f.write('\nSecond line')
```

```python
with open('file.txt', 'r') as f:
    i = 1
    for line in f:
        print(f"Line {i}: {line.strip()}")
        i += 1
```

Writing multiple lines:

```python
lines = ['Line 1\n', 'Line 2\n', 'Line 3\n']
with open('file.txt', 'w') as f:
    f.writelines(lines)
```

```python
with open('file.txt', 'r') as f:
    i = 1
    for line in f:
        print(f"Line {i}: {line.strip()}")
        i += 1
```

#Appending to files

### ✍️ Difference Between `'w'` (Write) and `'a'` (Append)

Both modes can **create a new file** if it doesn't exist, but they behave differently when the file **already exists**:

| Mode  | Behavior                                                                 |
|-------|--------------------------------------------------------------------------|
| `'w'` | Overwrites any existing content in the file.             |
| `'a'` | Adds new content to the **end** of the file, preserving existing data. |

✅ Use `'w'` when you want to start fresh, and `'a'` when you want to keep the existing content and just add more.


```python
with open('file.txt', 'a') as f:
    f.write('\nAppended line')
```

```python
with open('file.txt', 'r') as f:
    i = 1
    for line in f:
        print(f"Line {i}: {line.strip()}")
        i += 1
```

#Error Handling

```python
try:
    with open('nonexistent.txt', 'r') as f:
        content = f.read()
except FileNotFoundError:
    print("File not found")
except PermissionError:
    print("Permission denied")
except IOError:
    print("I/O error occurred")
```
