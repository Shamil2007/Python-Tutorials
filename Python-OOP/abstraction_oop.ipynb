#What Is Abstraction in Python

- Abstraction is the concept of hiding complex internal logic and showing only the essential features of an object.

it's like saying:

    “You don’t need to know how it works — just how to use it.”

In Python, abstraction is formally implemented using:

    ABC: Abstract Base Class

    @abstractmethod: marks required methods subclasses must implement

#Step 1: Define Abstract Base Class

```python
from abc import ABC, abstractmethod

class PaymentProcessor(ABC):
    @abstractmethod
    def authenticate(self):
        pass

    @abstractmethod
    def pay(self, amount):
        pass
```

🧠 What's happening?

- PaymentProcessor is the base class that defines a common interface.

- It contains two abstract methods:

    1. authenticate(): used to verify the user (e.g., login or card check)

    2. pay(amount): used to perform the actual payment

🔒 These methods must be implemented in subclasses. Otherwise, Python will raise an error.

#Step 2: Create Subclasses That Implement Abstract Methods

```python
class CreditCardPayment(PaymentProcessor):
    def authenticate(self):
        print("Verifying card with bank...")

    def pay(self, amount):
        print(f"Paid ${amount} using Credit Card.")

class PayPalPayment(PaymentProcessor):
    def authenticate(self):
        print("Logging in with PayPal credentials...")

    def pay(self, amount):
        print(f"Paid ${amount} using PayPal.")
```

#Step 3: Use It (Polymorphism + Abstraction)

```python
def process_payment(processor: PaymentProcessor, amount):
    processor.authenticate()
    processor.pay(amount)

# Use Credit Card
cc = CreditCardPayment()
process_payment(cc, 100)

# Use PayPal
pp = PayPalPayment()
process_payment(pp, 50)
```
