#Polymorphism – “Same Action, Different Behavior”

The Simple Idea
- Polymorphism means "many forms." It's like how different animals all "make sounds," but each makes a different sound.

```python
# All animals can make sounds, but each makes a different sound
class Animal:
    def make_sound(self):
        return "Some sound"

class Dog(Animal):
    def make_sound(self):  # Same method name, different action
        return "Woof!"

class Cat(Animal):
    def make_sound(self):  # Same method name, different action
        return "Meow!"

class Cow(Animal):
    def make_sound(self):  # Same method name, different action
        return "Moo!"

# The magic of polymorphism
animals = [Dog(), Cat(), Cow()]

for animal in animals:
    print(animal.make_sound())  # Same code, different results!
```

Why is This Useful?
- You can write code that works with many different types of objects without knowing exactly what type they are!

```python
def animal_concert(list_of_animals):
    print("🎵 Animal Concert Starting! 🎵")
    for animal in list_of_animals:
        print(f"♪ {animal.make_sound()} ♪")

# This function works with ANY animal!
zoo_animals = [Dog(), Cat(), Cow(), Dog()]
animal_concert(zoo_animals)
```

#Real-World Example: Shapes


```python
class Shape:
    def calculate_area(self):
        return 0  # Default implementation

class Rectangle(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def calculate_area(self):
        return self.width * self.height

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def calculate_area(self):
        return 3.14159 * self.radius * self.radius

# Same function works with different shapes!
def print_area(shape):
    area = shape.calculate_area()  # Don't care what type of shape it is!
    print(f"The area is: {area}")

# Using polymorphism
shapes = [
    Rectangle(5, 3),    # Area = 15
    Circle(4),          # Area = about 50.27
    Rectangle(2, 7)     # Area = 14
]

for shape in shapes:
    print_area(shape)
```
